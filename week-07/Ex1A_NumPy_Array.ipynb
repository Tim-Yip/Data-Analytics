# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array (
    [[11, 12, 13],
     [14, 15, 16]]
)
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The difference between the two loops is that the first loop goes through each row and the second loop goes through each column of each row. 

```python
for x in arrA:
    for y in x:
        loop_result = y * 2
print(loop_result)
```

This outputs 32 because at the end of the loop, the final value stored in y is 16 and the print statement is outside of the loop. 

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis = 0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis = 1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
print(stacked_array_2)
# This gives an error because the array shapes do not match.
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis = 1)
print(another_2)
# This gives a value error because the size at index 0 and index 1 do not match.
```

```python
stack1 = np.hstack((arrA, arrA)) # This stacks the arrays horizontally and places them next to each other.
print(stack1)
print()
stack2 = np.vstack((arrA, arrA)) # This stacks the arrays vertically and places one on top of the other.
print(stack2)
print()
stack3 = np.dstack((arrA, arrA)) # This stacks the arrays behind another creating a third dimension.
print(stack3)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2) # This splits into 2 2D arrays each with 2 rows and 3 columns.
print(splitsky)
```

```python
spliterino = np.array_split(joined_array_1, 3) # This splits into 3 2D arrays one with 2 rows and 3 columns and others with 1 row each.
print(spliterino)
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6], [13, 5], [7, 3]])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

```python
y = arrB >= 7
arrD = arrB[y]
print(arrD)
```
