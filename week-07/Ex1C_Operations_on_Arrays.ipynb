## Sample Arrays

## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

The rows got shuffled.

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print()
print(r_4)
```

If the original needs to be saved, use permutation, if not, use shuffle. 

## Mathematical Operations

```python
print(r_1)
print(r_1.shape)
print()

print(r_2)
print(r_2.shape)
print()

result = np.add(r_1, r_2)

print(result)
print(result.shape)
```

```python
print(r_1)
print(r_1.shape)
print()

print(r_3)
print(r_3.shape)
print()

result = np.add(r_1, r_3)

print(result)
print(result.shape)
```

```python
print(r_2)
print(r_2.shape)
print()

print(r_4)
print(r_4.shape)
print()

print(np.add(r_2, r_4))

# There is a value error because both arrays have different shapes and cannot be added together.
```

```python
print(np.add(np.array([1]), r_4))
```

```python
print(np.add(np.array([[1]]), r_4))
```

```python
print(np.add(np.array([1, 1, 1, 1, 1]), r_4))
```

```python
print(np.add(np.array([[1], [1]]), r_4))
# ValueError: operands could not be broadcast together with shapes (2,1) (4,5) 
```

Broadcasting works by performing opperations on arrays of different shapes. When using add, corresponding elements are added together. If the dimensions don't match, it gives a value error.

## Grocery Problem

```python
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])

print(bag1)
print(bag2)
```

```python
bag1_total = np.sum(bag1)
bag2_total = np.sum(bag2)

print(f"The first bag contains {bag1_total} items, and the second bag contains {bag2_total} items.")
```

```python
category_total = np.add(bag1, bag2)

print(f"There are {category_total[0]} vegetables, {category_total[1]} fruits, {category_total[2]} boxes of snacks, "
f"and {category_total[3]} beverages in the two grocery bags.")
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])
```

```python
ben = np.array([0, 5, 5, 3, 4])
```

```python
carla = np.array([3, 6, 0, 3, 4])
```

```python
all_knocks = np.vstack((ali, ben, carla))

print(all_knocks)
```

```python
individual_totals = np.sum(all_knocks, axis = 1)
group_total = np.sum(all_knocks)

print(f"Ali: {individual_totals[0]} doors.")
print(f"Ben: {individual_totals[1]} doors.")
print(f"Carla: {individual_totals[2]} doors.")
print(f"Total: {group_total} doors.")
```

```python
daily_totals = np.sum(all_knocks, axis = 0)
rolling_total = np.cumsum(daily_totals)
estimated_votes = rolling_total * 0.5

print(estimated_votes)
```
