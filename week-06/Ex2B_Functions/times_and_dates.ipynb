Working with datetime functions.

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
```

```python
day_week = today.strftime("%A")
month = today.strftime("%B")
day_month = today.strftime("%d")
year = today.strftime("%Y")

print(f"Today is {day_week}, {month} {day_month}, {year}")
```

# My birthday

```python
birthday = datetime.date(2026, 7, 5)
```

```python
print(f"My birthday is {birthday.strftime('%m/%d/%Y')}")
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)
```

```python
print(f"90 days from today is {ninety_d.strftime('%B' + " " + '%d' + "," + '%Y')}")
```

# Time for Dinner

```python
dinner_time = datetime.time(18, 30)
```

```python
print(f"Let's meet for dinner at {dinner_time.strftime('%I:%M %p')}")
```
