# EX2C Lab3 Tax Functions
Using functions to calculate payroll taxes.

```python
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062
```

```python
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145
```

```python
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        tax_rate = 0.23
    elif withholding_code == 1:
        tax_rate = 0.21
    elif withholding_code == 2:
        tax_rate = 0.195
    elif withholding_code == 3:
        tax_rate = 0.185
    else:
        tax_rate = 0.18 - (withholding_code - 4) * 0.005

    return gross_pay * tax_rate
```

```python
gross_pay = 750
withholding_code = 0

print(f"{'Social Security Tax:':<22}${get_soc_sec_tax(gross_pay):.2f}")
print(f"{'Medicare Tax:':<22}${get_medicare_tax(gross_pay):.2f}")
print(f"{'Federal Tax:':<22}${get_federal_tax(gross_pay, withholding_code):.2f}")
```
