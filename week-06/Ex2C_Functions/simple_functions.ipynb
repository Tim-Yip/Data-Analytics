Working with defining custom functions.

```python
def favorite_things(name, movie, music):
    return(f"Name: {name}\nFavorite movie: {movie}\nFavorite musician: {music}")

print(favorite_things("Timothy", "Totoro", "Illenium"))
```

```python
def why_im_here(purpose):
    return(f"I joined the Year Up United Data Analyst program because {purpose}")

print(why_im_here("I wanted to start building my career and develop new skills."))
```

```python
def favorite_place(place, why):
    return(f"One of my favorite places is {place} because {why}")
```

```python
print(favorite_place("Singapore", "the food is very good while being cheap and it's a fun place to explore."))
```

```python
print(favorite_place("home", "home is comfy."))
```
